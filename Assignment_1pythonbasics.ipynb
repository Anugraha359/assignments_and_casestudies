1. Write a program that takes the user's account balance and withdrawal amount. If 
sufficient balance is available, deduct the amount and display the updated balance. 
Otherwise, print an error message. 
Sample Input: 
Balance = 5000
Withdrawal  = 1500 
Sample Output: 
Transaction successful. Remaining balance: ₹3500 

```python
print("Simple ATM Calculator")
#input from the user
balance_amount=eval(input("Enter the account balance: "))
withdrawal_amount=eval(input("Enter the withdrawal bal: "))

#checking balance
if withdrawal_amount <= balance_amount:
    balance_amount = balance_amount-withdrawal_amount
    #display updated balance
    print("Transaction successful. Remaining balance: ₹",balance_amount)

else:
    print("Error: Insufficient balance")
```

2. Electricity Bill Generator Create a bill calculator where the user enters the number of units consumed. The rate is ₹5 per unit for the first 100 units, ₹7 for the next 100, and ₹10 for anything above 200 units. Calculate and display the total bill. Sample Input: Units consumed=275 Sample Output:
Bill amount = ₹1000 (100×5 + 100×7 + 75×10)

```python
print("Electricity Bill Generator ")
#input from user
units_consumed = eval(input("Enters the number of units consumed"))
bill = 0
#checking bill amount
if(units_consumed <= 100):
    bill = units_consumed*5
elif(units_consumed <= 200):
    bill = (100*5)+((units_consumed-100)*7)
else:
    bill = (100*5)+(100*7)+((units_consumed-200)*10)
    
#display bill amount
print("Bill Amount",bill)
```

3. Temperature Converter 
Write a Python program to convert a given temperature value between Celsius and 
Fahrenheit. Accept input as a number and a unit indicator ('C' or 'F'). 
Sample Input: 
Temp  = "37"
Unit     "c"
Sample Output:
Temperature in Fahrenheit: 98.6°F

```python
#input from user
temp_value = eval(input("Enter the tempature value:"))
unit = input("Enter the unit (C/F)")
#conversion
if unit == "C":
    fahernheit = (temp_value * 9/5) + 32
    print("Temperature in Fahernheit",fahernheit,"°F")

elif unit == "F":
    celsius = (temp_value - 32) * 5/9
    print("Temperature in Celsius",celsius,"°C")

else:
    print("Invalid unit, Please enter C or F as unit.")
                
```

 4.  Mobile Recharge Validity 
Write a program that takes the recharge date and today’s date as input and checks 
whether the 28-day recharge plan is still valid or expired. 
Sample Input: 
Recharge date: 2025-07-01
Today :  2025-07-30 
Sample Output: 
Plan expired. Recharge again.

```python
from datetime import datetime, timedelta

#input from user
recharge_date_str = input("Enter the last recharge date:")
today_date_str = input("Enter today's date:")

# Convert string to date
recharge_date = datetime.strptime(recharge_date_str, "%Y-%m-%d")
today_date = datetime.strptime(today_date_str, "%Y-%m-%d")

#validity check
validity_end_date = recharge_date + timedelta(days=28)

#check validity
if today_date <= validity_end_date:
    print("Plan is still valid.")
else:
    print("Plan expired. Recharge again.")
```

5. Bus Seat Reservation System: 
A list contains the names of passengers in order of booking. A cancellation occurs; remove 
the name from the list and shift the rest. Also, insert a new booking at a specified position. 
Sample Input:
Booked = ['Amit', 'Sara', 'John'], 
Cancel = 'Sara', 
New = 'Neha', Position = 1
Sample Output: 
['Amit', 'Neha', 'John']

```python
print("Bus Seat Reservation System")
#input from user
passenger_name=['Amit','Sara','John','Amay']
print("first passenger list:",passenger_name)

#inputs
cancel="Sara"
new_booking="Neha"
position=1 #index position
#Remove cancelled passenger
if cancel in passenger_name:
    passenger_name.remove(cancel)

#insert new booking at given position
passenger_name.insert(position, new_booking)

#final list
print("Updated passenger list: ",passenger_name)
```

6.  Sum of even numbers 
Write a program that takes a number as input and prints the sum of all even numbers 
from 1 to that number. 
Sample Input: 
Number: 10 
Sample Output: 
Sum of even numbers: 30 
(2 + 4 + 6 + 8 + 10)

```python
#user input
print("Sum of even numbers ")
limit=int(input("enter the limit: "))
even_numbers=[]
sum=0
for i in range(1,limit+1):
    if i%2 == 0:
        even_numbers.append(str(i))
        sum = sum + i
print("Sum of even numbers:",sum)
print("(" + " + ".join(even_numbers) + ")")
```

6. Dictionary – Student Marks Summary 
Write a program that takes input of 3 students’ names and their marks, stores them in a 
dictionary, and prints the average marks. 
Sample Input: 
Enter name: Alice   
Enter marks: 80   
Enter name: Bob   
Enter marks: 90   
Enter name: Charlie   
Enter marks: 70 

Sample Output: 
Average Marks: 80.0 

```python
print(" Dictionary – Student Marks Summary")
students={}
n=int(input("enter the no.of students"))
sum=0

#input
for i in range(n):
    name = input("Enter name: ")
    marks = int(input("Enter marks: "))
    students[name] = marks
    sum += marks

#calculate avarege
average = sum/n

#output
print("Average Marks: ",average)
```

8. Dictionary – Word Frequency 
Write a program that takes a sentence and displays the frequency of each word using a 
dictionary. 
Sample Input: 
Sentence: hello world hello python 
Sample Output: 
hello: 2   
world: 1   
python: 1

```python
#input sentence
sentence = input("Enter a sentence: ")

#split the sentence into words
words = sentence.split()

#create empty dictionary
word_frequency = {}

#count frequency of words
for word in words:
    if word in word_frequency:
        word_frequency[word] += 1
    else:
        word_frequency[word] = 1

#display the frequency
for word, count in word_frequency.items():
    print(f"{word}: {count}")
```

9.  Vowel Counter 
Write a function that takes a string and returns the number of vowels in it.  
Sample Input: 
Enter a string: Python Programming 
Sample output: 
Number of vowels: 4

```python
#function declaration
def counting_vowels(text):
    vowels="AEIOUaeiou"
    count=0

    for i in text:
        if i in vowels:
           count = count+1

    return count


#input
string = input("Enter a string: ")

#function call
result = counting_vowels(string)

#output
print("Number of vowels:", result)
```

Write a program that takes two numbers as input: a range limit N and a divisor D. Count 
how many numbers between 1 and N are divisible by D.
Sample Input: 
Enter the range limit (N): 50   
Enter the divisor (D): 5 
Sample output: 
Count of numbers divisible by 5: 10 
(i.e., 5, 10, 15, ..., 50) 

```python
#input
N = int(input("Enter the range limit (N): "))
D = int(input("Enter the divisor (D): "))

count = 0
numbers = []

# Loop from 1 to N
for i in range(1, N + 1):
    if i % D == 0:
        count += 1
        numbers.append(str(i))

# Output
print(f"Count of numbers divisible by {D}: {count}")
print("(" + ", ".join(numbers) + ")")
```
